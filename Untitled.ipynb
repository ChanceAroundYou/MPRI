```python
import process
import os
from DM.file_manage import RotatedNiiFileManager, LabelNiiFileManager
import numpy as np
import matplotlib.pyplot as plt
import cv2
from scipy.ndimage import rotate as scipy_rotate
```

```python
def clear_scp(scp):
    max_low = 0
    
    if np.where(scp)[1].min() < 120:
        x_range = range(0, int(0.7*np.where(scp)[1].max()))
    else:
        x_range = range(scp.shape[1]-1, int(1.1*np.where(scp)[1].min()), -1)
    for x in x_range:
        col = scp[:, x]
        if col.any():
            low = np.where(col == 0)[0].max()
            if low > max_low and low < 129:
                max_low = low
                max_x = x
    if np.where(scp)[1].min() > 120:
        scp[:, max_x:] = 0
    else:
        scp[:, :max_x] = 0
    return scp
```

```python
def get_scp_num(img_nii, r_quad, mid, size=10):
    start = r_quad[1] - 1
    center_y = r_quad[0] * size
    center_x = mid * size
    resize_shape = (img_nii.size[0] * size, img_nii.size[2] * size)
    box = (
        slice(center_y-20, center_y+150),
        slice(center_x-130, center_x+130)
    )
    for i in range(5):
        img = img_nii.get_slice(start + i, dim=1)
        resized = cv2.resize(img, resize_shape, interpolation=cv2.INTER_CUBIC)
        part_img = resized[box]
        ot = process.get_otsu(part_img)
        mask = process.get_bin_image(part_img, ot*1.25)
        _, label, stats, centers = cv2.connectedComponentsWithStats(mask, connectivity=4)
        low_components = [(num, stat) for num, stat in enumerate(stats) if stat[1]+stat[3] > 100 and num > 0]
        low_components = sorted(low_components, key=lambda x: x[1][1])
#         print(low_components)
        if low_components[0][1][1] > 20:
            return resized, center_y, center_x
```
