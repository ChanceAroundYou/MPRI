```python
base_path = 'data/accubrainresult/01-03/PA2'
img_path = os.path.join(base_path, 'T1_rigid.nii.gz')
label_path = os.path.join(base_path, 'result_seg_rigid.nii.gz')
img_nii = RotatedNiiFileManager(img_path)
label_nii = LabelNiiFileManager(label_path)
img_nii.load()
label_nii.load()
img_nii.normalize()
```

```python
quad, mid, _, _, ang, r_quad = process.first_stage.run(img_nii, label_nii)
plt.imshow(process.add_points(img_nii.get_slice(mid, dim=2), quad, 2))
```

```python
mid_label = label_nii.get_slice(mid, dim=2)
ang = process.first_stage.get_brainstem_angle(mid_label)
img_nii.rotate(ang)
label_nii.rotate(ang)

plt.imshow(process.add_points(img_nii.get_slice(mid, dim=2), r_quad, 2))
```

```python
img = img_nii.get_slice(r_quad[1], dim=1)
plt.figure(figsize=(10, 10))
plt.imshow(img)
```

```python
part_img = img[r_quad[0]+3:r_quad[0]+15, mid-9:mid+9]
plt.imshow(part_img)
```

```python
cl_img = process.get_clahe_image(part_img, 0.03, 5, 4)
ot = process.get_otsu(cl_img)
plt.imshow(cl_img)
```

```python
mask = process.get_bin_image(cl_img, ot*1.4)
plt.imshow(mask)
```

```python
_, label, stats, centers = cv2.connectedComponentsWithStats(mask, connectivity=4)

buttom_components = [(num, stat) for num, stat in enumerate(stats) if stat[1]+stat[3] > 10 and num > 0]
scp_components = sorted(buttom_components, key=lambda x: x[1][4], reverse=True)

plt.imshow(np.logical_or(label==scp_components[0][0], label==scp_components[1][0]))
scp_components
```

```python
cv2.fitEllipse(np.stack(np.where(label==scp_components[0][0]), axis=1))
```
