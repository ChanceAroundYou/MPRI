```python
import process
import os
from DM.file_manage import RotatedNiiFileManager, LabelNiiFileManager
import numpy as np
import matplotlib.pyplot as plt
import cv2
from scipy.ndimage import rotate as scipy_rotate
```

```python
base_path = 'data/accubrainresult/01-03/PA0'
img_path = os.path.join(base_path, 'T1_rigid.nii.gz')
label_path = os.path.join(base_path, 'result_seg_rigid.nii.gz')
img_nii = RotatedNiiFileManager(img_path)
label_nii = LabelNiiFileManager(label_path)
img_nii.load()
label_nii.load()
img_nii.normalize()

quad, mid, _, _, ang, r_quad = process.first_stage.run(img_nii, label_nii)
```

```python
process.third_stage.run(img_nii, label_nii, r_quad, mid, ang)
```

```python
# ang += np.deg2rad(8)
# r_quad = process.first_stage.get_rotated_quad_seg_point(r_quad, mid_label.shape, np.deg2rad(8))
img_nii.rotate(ang)
label_nii.rotate(ang)
# plt.figure(figsize=(10, 10))
# plt.imshow(process.add_points(img_nii.get_slice(mid, dim=2), r_quad, 2))

size = 10
resized, center_y, center_x = process.third_stage.get_scp_slice(img_nii, r_quad, mid, size)

# img = img_nii.get_slice(num, dim=1)
# size = 10
# ori_shape = img.shape
# resize_shape = tuple(map(lambda x: x*size, ori_shape))
# resized = cv2.resize(img, resize_shape, interpolation=cv2.INTER_CUBIC)
# plt.figure(figsize=(10, 10))
# plt.imshow(resized)
# plt.imshow(img_nii.get_slice(r_quad[1], dim=1))

box = [-10, 150, -130, 130]
part_img = resized[center_y+box[0]:center_y+box[1], center_x+box[2]:center_x+box[3]]
# cl_img = process.get_clahe_image(part_img, 0.03, 8, 8)
# ot = process.get_otsu(cl_img)
# plt.imshow(cl_img)
ot = process.get_otsu(part_img)
mask = process.get_bin_image(part_img, ot*1.26)
# plt.imshow(mask)

_, label, stats, centers = cv2.connectedComponentsWithStats(mask, connectivity=4)

buttom_components = [
    (num, stat) for num, stat in enumerate(stats)
    if stat[1]+stat[3] > 130 and num > 0 and (110 <stat[0] < 180 or 80 < stat[0]+stat[2] < 150)]
scp_components = sorted(buttom_components, key=lambda x: x[1][4], reverse=True)
scp_0 = process.third_stage.get_clear_scp((label==scp_components[0][0]).astype(np.uint8))
scp_1 = process.third_stage.get_clear_scp((label==scp_components[1][0]).astype(np.uint8))
scp_img = scp_0 + scp_1
plt.imshow(scp_img)
```
